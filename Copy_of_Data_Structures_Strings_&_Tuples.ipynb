```python
# String Concatenation

name = input("Enter your Name: ")

string1 = "Hello "
string2 = name
string3 = ", welcome to Python programming"

concatenated_string = string1 + string2 + string3

print(concatenated_string)
```

```python
# String Slicing and Indexing

name = input("Enter your Name: ")

string1 = "Hello "
string2 = name
string3 = ", welcome to Python programming"

concatenated_string = string1 + string2 + string3

print("Concatenated String:", concatenated_string)
```

```python
# a. Print the first character

print("First character:", concatenated_string[0])
```

```python
# b. Print the last character

print("Last character:", concatenated_string[-1])
```

```python
# c. Print the first 5 characters

print("First 5 characters:", concatenated_string[:5])
```

```python
# d. Print the last 11 characters

print("Last 11 characters:", concatenated_string[-11:])
```

```python
# e. Print the string in reverse

print("Reversed string:", concatenated_string[::-1])
```

```python
# f. Print the word "Python"

start = concatenated_string.find("Python")
print("Word Python:", concatenated_string[start:start+6])
```

```python
# String Methods

strM = "Python beginner tutorial"

# a. Convert the sentence to uppercase

print("Uppercase:", strM.upper())
```

```python
# c. Capitalize the sentence

print("Capitalized:", strM.capitalize())
```

```python
# d. Count the occurrences of character 't'

print("Number of 't' occurrences:", strM.count("t"))
```

```python
# e. Replace "Python" with "Data Analytics"

result = strM.replace("Python", "Data Analytics")

print("After replacement:", result)
```

```python
# Tuples

tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

print("First tuple:", tuple1)
print("Second tuple:", tuple2)
```

```python
# a. Concatenate the two tuples

t_combine = tuple1 + tuple2

print("Combined tuple:", t_combine)
```

```python
# b. Repeat the elements of t_combine 3 times

t_repeat = t_combine * 3

print("Repeated tuple:", t_repeat)
```

```python
# c. Access the 3rd element from t_combine

print("3rd element:", t_combine[2])
```

```python
# d. Access the first three elements from t_combine

print("First three elements:", t_combine[:3])
```

```python
# e. Access the last three elements from t_combine

print("Last three elements:", t_combine[-3:])
```
